```python
from pypdf import PdfReader, PdfWriter

def merge_pdfs(pdf_list, output_path):
    writer = PdfWriter()

    for pdf_file in pdf_list:
        reader = PdfReader(pdf_file)
        for page in reader.pages:
            writer.add_page(page)

    with open(output_path, "wb") as f_out:
        writer.write(f_out)

pdf_files = ['/content/English-1.pdf', '/content/English-2.pdf']
output_file = 'ENG_Merge.pdf'

merge_pdfs(pdf_files, output_file)
```

```python
import fitz
import re
from fpdf import FPDF


def clean_text(text):
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

class PDF(FPDF):
    def header(self):
        self.set_font("Arial", "B", 12)

    def chapter_body(self, text):
        self.set_font("Arial", "", 11)
        self.multi_cell(0, 10, text)

# Extract text, clean it, and write to PDF
def create_cleaned_pdf(input_pdf_path, output_pdf_path):
    doc = fitz.open(input_pdf_path)
    cleaned_text = ""
    for page in doc:
        text = page.get_text()
        cleaned_text += clean_text(text) + "\n\n"

    # PDF
    pdf = PDF()
    pdf.add_page()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.chapter_body(cleaned_text)
    pdf.output(output_pdf_path)
    print(f"Cleaned PDF saved as: {output_pdf_path}")
create_cleaned_pdf("ENG_Merge.pdf", "Cleaned_English.pdf")
```
